# Simple Linear Regression

```python
#Import the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from pandas.core.common import random_state
from sklearn.linear_model import LinearRegression
```

```python
#Load the dataset
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Laboratoare_Învățare_automată/Lab1/train.csv')
df.head(10)
```

```python
#Info about dataset
df.info()
df.describe()
```

```python
#Data Cleaning
print(df.isnull().sum())
print("Max of missing values:", df.isnull().sum().max())
print("Number of duplicated rows:", df.duplicated().sum())
```

```python
#Splitting the features
X = df[['LotArea']]              #independent feature
y = df[['SalePrice']]            #dependent feature

#Visualising the plot
plt.scatter(X, y, color = 'green')
plt.title('House Price')
plt.xlabel('LotArea')
plt.ylabel('SalePrice')
plt.show()
```

```python
#Splitting dataset into train/test dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)

#Training
model = LinearRegression()
model.fit(X_train, y_train)

#Prediction
y_pred = model.predict(X_test)

print(f"Intercept(w0): {model.intercept_}")
print(f"Slope(w1): {model.coef_}")
#y= 161179.01 + 1.85X

```

```python
#Visualising the prediction
plt.scatter(X_train, y_train, color = 'green')
plt.plot(X_test, y_pred, color = 'purple')
plt.title('House Price')
plt.xlabel('LotArea')
plt.ylabel('SalePrice')
plt.show()
```

```python
#Evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```
