```python
import pandas as pd
import warnings
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
warnings.filterwarnings("ignore", category=DeprecationWarning)
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Laboratoare_Învățare_automată/Lab5/Groceries_dataset.csv")
df.head(10)
```

```python
df.info()
```

```python
num_items = df['itemDescription'].nunique()
print("Number of distinct items:", num_items)
```

```python
df['itemDescription'].unique()
```

```python
num_customers = df['Member_number'].nunique()
print("Number of distinct customers:", num_customers)
```

```python
df['Date'] = pd.to_datetime(df['Date'], utc = True)
df['Date'].min(),df['Date'].max()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
transactions = df.groupby(['Member_number', 'Date'])['itemDescription'].apply(list)
```

```python
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
transactions_matrix = pd.DataFrame(te_ary, columns=te.columns_)
```

```python
transactions_matrix.head()
```

```python
transactions_matrix.shape
```

```python
frequent_items = apriori(transactions_matrix, min_support=0.002, use_colnames=True)
frequent_items.head(10)
```

```python
rules = association_rules(frequent_items, metric="confidence", min_threshold=0.01)
rules = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
rules.head(10)
```

```python
import matplotlib.pyplot as plt
import numpy as np

norm = plt.Normalize(rules['confidence'].min(), rules['confidence'].max())
cmap = plt.get_cmap("viridis")

plt.figure(figsize=(14, 6))
scatter = plt.scatter(rules['confidence'], rules['lift'], c=rules['confidence'], cmap=cmap, alpha=0.6)


for i in range(len(rules)):
    plt.annotate('', (rules['confidence'][i], rules['lift'][i]), fontsize=8, alpha=0.7)

plt.xlabel('Confidence')
plt.ylabel('Lift')
plt.title('Association Rules Confidence vs Lift')

plt.subplots_adjust(right=0.80)

cbar = plt.colorbar(scatter, fraction=0.02, pad=0.04)
cbar.set_label('Confidence')
plt.text(1.3, 1, "Association Rules:", ha='left', va='top', fontsize=12, transform=plt.gca().transAxes)
for i in range(len(rules)):
    rule_text = f"{list(rules['antecedents'][i])} -> {list(rules['consequents'][i])}\n" \
                f"Conf: {rules['confidence'][i]:.2f}, Lift: {rules['lift'][i]:.2f}"
    color = cmap(norm(rules['confidence'][i]))
    plt.text(1.3, 0.9 - i * 0.1, rule_text, ha='left', va='top', fontsize=10, transform=plt.gca().transAxes, color=color)
plt.show()
```
