```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import time
```

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Laboratoare_Învățare_automată/Lab7/gym_members_exercise_tracking.csv")
```

```python
df = pd.get_dummies(df, columns=['Gender'], drop_first=True)
le = LabelEncoder()
df['Workout_Type_Encoded'] = le.fit_transform(df['Workout_Type'])
```

```python
X = df.drop(['Workout_Type', 'Workout_Type_Encoded'], axis=1)
y = df['Workout_Type_Encoded']
X_numeric = X.select_dtypes(include=['number'])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_numeric, y, test_size=0.3, random_state=42, stratify=y)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
results = {}
model_name_lr = "Logistic Regression (Clasic)"
start_time_lr = time.time()
lr_model = LogisticRegression(max_iter=1000, random_state=42)
lr_model.fit(X_train_scaled, y_train)
end_time_lr = time.time()
```

```python
y_pred_lr = lr_model.predict(X_test_scaled)
accuracy_lr = accuracy_score(y_test, y_pred_lr)
training_time_lr = end_time_lr - start_time_lr
results[model_name_lr] = {'Accuracy': accuracy_lr, 'Training Time (s)': training_time_lr}
```

```python
model_name_rf = "Random Forest (Ensemble)"
start_time_rf = time.time()
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train_scaled, y_train)
end_time_rf = time.time()
```

```python
y_pred_rf = rf_model.predict(X_test_scaled)
accuracy_rf = accuracy_score(y_test, y_pred_rf)
training_time_rf = end_time_rf - start_time_rf
results[model_name_rf] = {'Accuracy': accuracy_rf, 'Training Time (s)': training_time_rf}
```

```python
results_df = pd.DataFrame.from_dict(results, orient='index')
print("\nComparison Results:")
print(results_df.sort_values(by='Accuracy', ascending=False))
```
